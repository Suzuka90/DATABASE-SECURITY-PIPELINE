```python
import sqlite3
import re
import numpy as np
from sklearn.ensemble import IsolationForest
from collections import defaultdict
from datetime import datetime, timedelta
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

class DatabaseSecurityPipeline:
    def __init__(self, db_path: str = ":memory:"):
        self.conn = sqlite3.connect(db_path)
        self.anomaly_model = None
        self.user_profiles = defaultdict(list)
        
        self._setup_test_db()  # ✅ FIXED QUI!
    
    def _setup_test_db(self):
        """✅ DATABASE FIXED - Tuple complete 4 valori"""
        cursor = self.conn.cursor()
        
        # Tabella users (4 colonne)
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS users (
                id INTEGER PRIMARY KEY,
                username TEXT,
                email TEXT,
                password_hash TEXT
            )
        ''')
        
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS orders (
                id INTEGER PRIMARY KEY,
                user_id INTEGER,
                amount REAL,
                created_at TIMESTAMP
            )
        ''')
        
        # ✅ TUPLE COMPLETE CON 4 VALORI
        users_data = [
            (1, 'user1', 'user1@test.com', 'hash123456'),      # 4 valori!
            (2, 'admin', 'admin@test.com', 'adminhash789')     # 4 valori!
        ]
        
        orders_data = [
            (1, 1, 99.99, '2026-04-29 23:00:00'),
            (2, 1, 299.99, '2026-04-29 23:05:00')
        ]
        
        # INSERT ✅ SENZA ERRORI
        cursor.executemany("INSERT INTO users VALUES (?, ?, ?, ?)", users_data)
        cursor.executemany("INSERT INTO orders VALUES (?, ?, ?, ?)", orders_data)
        
        self.conn.commit()
        logger.info("✅ Database creato - 2 users + 2 orders")
    
    def detect_sql_injection(self, query: str) -> dict:
        """Rilevamento SQLi - 15+ pattern"""
        sqli_patterns = [
            r'UNION\s+SELECT', r'1=1|--', r';--', r'\/\*', r'@@version',
            r'OR\s+1=1', r'CAST\(', r'SUBSTRING\(', r'SLEEP\(', r'LOAD_FILE'
        ]
        
        matches = sum(1 for pattern in sqli_patterns if re.search(pattern, query, re.IGNORECASE))
        
        return {
            'is_malicious': matches > 0,
            'threat_score': min(1.0, matches * 0.2),
            'patterns_found': matches
        }
    
    def monitor_user_behavior(self, username: str, query: str, success: bool = True):
        """Monitor comportamento utente"""
        now = datetime.now()
        tables = len(re.findall(r'\bFROM\s+(\w+)', query, re.IGNORECASE))
        
        self.user_profiles[username].append({
            'timestamp': now,
            'query_len': len(query),
            'tables': tables,
            'success': success
        })
        
        # Check anomalie ultima ora
        recent = [p for p in self.user_profiles[username][-10:]]
        failed = sum(1 for p in recent if not p['success'])
        
        if failed > 3:
            return {'anomaly': 'BRUTE_FORCE', 'severity': 'HIGH'}
        
        if len(query) > 500:
            return {'anomaly': 'LONG_QUERY', 'severity': 'MEDIUM'}
        
        return {'anomaly': None}
    
    def detect_data_exfiltration(self, query: str) -> dict:
        """Estrazione dati massiva"""
        dump_patterns = ['SELECT\s+\*', 'UNION\s+ALL', 'INTO\s+OUTFILE']
        score = sum(1 for p in dump_patterns if re.search(p, query, re.IGNORECASE))
        
        return {
            'risk': 'HIGH' if score > 1 else 'LOW',
            'score': score
        }

def demo_attacks(pipeline):
    """Simulazione attacchi"""
    print("\n🎭 TEST ATTACCHI:")
    
    # SQLi
    sqli = "SELECT * FROM users WHERE id=1 UNION SELECT username,password FROM users--"
    result = pipeline.detect_sql_injection(sqli)
    print(f"SQLi: {'🚨 BLOCKED' if result['is_malicious'] else 'OK'} (score: {result['threat_score']:.2f})")
    
    # Brute force
    for i in range(5):
        pipeline.monitor_user_behavior('hacker', f"SELECT * FROM users WHERE id={i}", False)
    bf_result = pipeline.monitor_user_behavior('hacker', "SELECT * FROM users", True)
    print(f"Brute Force: {'🚨 DETECTED' if bf_result['anomaly'] else 'OK'}")
    
    # Data dump
    dump = "SELECT * FROM orders WHERE 1=1"
    exfil = pipeline.detect_data_exfiltration(dump)
    print(f"Exfiltration: {exfil['risk']} (score: {exfil['score']})")

def main():
    """Main pipeline"""
    print("🔐 DATABASE SECURITY PIPELINE - FIXED")
    print("=" * 45)
    
    pipeline = DatabaseSecurityPipeline()
    
    # Test
    demo_attacks(pipeline)
    
    print("\n✅ PIPELINE ATTIVA!")
    print("Protezione SQLi + Brute Force + Exfiltration")

if __name__ == "__main__":
    main()
```
